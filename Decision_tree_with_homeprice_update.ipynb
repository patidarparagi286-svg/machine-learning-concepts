```python
X=df.drop(columns="sales")
y=df['sales']
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('scale', StandardScaler(), X.columns)    #scale all feature columns
    ],
    remainder="drop"   #drop all other columns not listed in transformers
  )
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.4,random_state=42)
```

```python
#model : decision tree(infomation gain method = entropy(default) , gini)
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=2,
    random_state=42
)
```

```python
#Pipeline : preprocessing + maodel
pipe = Pipeline(steps=[
    ('prep', preprocessor),
    ('model', model)
])
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
print(accuracy_score(y_test, y_pred))
```
