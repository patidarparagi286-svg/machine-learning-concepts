```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

```

```python
iris=load_iris()
X=iris.data
print("Iris data shape:",X)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
wcss=[]
K_value=range(1,11)

for k in K_value:
    km=KMeans(n_clusters=k,random_state=42)
    km.fit(X_scaled)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.figure()
plt.plot((K_value),wcss, marker='o')
plt.title("Elbow curve(k vs WCSS/Inertia)")
plt.xlabel("number of clusters(k))")
plt.ylabel("WCSS(Inertia)")

plt.show()
```

```python
sil_score=[]

for k in range(2,11):
  km = KMeans(n_clusters=k, random_state=42)
  labels=km.fit_predict(X_scaled)
  score=silhouette_score(X_scaled,labels)
  sil_score.append(score)
  print(f'K={k}: Silhouette={score:.3f}')
```

```python
plt.figure()
plt.plot(range(2,11),sil_score,marker='o')
plt.title("Silhouette score vs number of clusters")
plt.xlabel("number of clusters(k)")
plt.ylabel("Silhouette score")
plt.show()
```

```python
kmeans = KMeans(n_clusters=2,random_state=42,n_init=10)
final_labels=kmeans.fit_predict(X_scaled)
```

```python
final_labels
```

```python
print('\nCluster counts:')
for c in range(2):
  print(f'Cluster {c}: {(final_labels==c).sum()}data points')
```

```python
#PCA - Principal Component Analysis (used to reduce unwanted columns of dataset)
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)
X_2d
```

```python
plt.figure()
plt.scatter(X_2d[:,0],X_2d[:,1],c=final_labels)
plt.title(f'Iris clusters using KMeans(k={2})-PCA 2D View')
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.show()
```
